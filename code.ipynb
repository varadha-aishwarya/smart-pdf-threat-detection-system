**Removing Outlier**

```python
# Get the numerical columns of the DataFrame
numerical_columns = df.select_dtypes(include=['float64', 'int64']).columns

# Define the function to remove outliers using the IQR method
def remove_outliers_iqr(column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df[column] = df[column].clip(lower=lower_bound, upper=upper_bound)

# Iterate over each numerical column and remove outliers
for column in numerical_columns:
    remove_outliers_iqr(column)
```

**DNN**

```python
# Ensure x_train and y_train are NumPy arrays
X_train = np.asarray(x_train)
X_test = np.asarray(x_test)
Y_train = np.asarray(y_train)
Y_test = np.asarray(y_test)

# Define the model
model = Sequential()

# Add layers to the model
# Input X_train
model.add(Dense(128, input_shape=(X_train.shape[1],), activation='relu'))

# Hidden layers
model.add(Dense(64, activation='relu'))
model.add(Dropout(0.5))  # Dropout to prevent overfitting
model.add(Dense(32, activation='relu'))

# Output layer
model.add(Dense(1, activation='sigmoid'))  # Assuming binary classification, use 'softmax' for multiclass

# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Train the model
model.fit(X_train, Y_train, epochs=10, batch_size=32, validation_split=0.2)
```

```python
# Make predictions (for binary classification, round the predictions to get class labels)
y_pred = model.predict(X_train)
y_pred = np.round(y_pred)

# Calculate the metrics
acc_dnn = accuracy_score(Y_train, y_pred)
ps_dnn = precision_score(Y_train, y_pred)
rs_dnn = recall_score(Y_train, y_pred)
f1_dnn = f1_score(Y_train, y_pred)

# Print the scores
print(f'Accuracy Score = {acc_dnn}')
print(f'Precision Score = {ps_dnn}')
print(f'Recall Score = {rs_dnn}')
print(f'F1 Score = {f1_dnn}')
```
